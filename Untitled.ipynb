```python
# youtube seconds conversion
def seconds_to_timestamp(seconds):

    h = int(seconds // 3600)
    m = int((seconds % 3600) // 60)
    s = int(seconds % 60)

    if h > 0:
        return f"{h}:{m}:{s}"

    return f"{m}:{s}"
# // is floor value division
# / is simple division
```

```python
from langchain_core.runnables import RunnableParallel, RunnablePassthrough,RunnableLambda
from langchain_core.output_parsers import StrOutputParser
```

```python
# def format_docs(retrieved_docs):
#   context_text = "\n\n".join(doc.page_content for doc in retrieved_docs)
#   return context_text
```

```python
def seconds_to_timestamp(seconds):
    h = seconds // 3600
    m = (seconds % 3600) // 60
    s = seconds % 60

    if h > 0:
        return f"{h:02}:{m:02}:{s:02}"

    return f"{m:02}:{s:02}"


def format_docs(docs):
    formatted = []

    for doc in docs:
        timestamp = seconds_to_timestamp(int(doc.metadata["start"]))

        formatted.append(
            f"[Timestamp: {timestamp}]\n{doc.page_content}"
        )

    return "\n\n".join(formatted)
```

```python
parser=StrOutputParser()
```

```python
parallel_chain= RunnableParallel({
    "question": RunnablePassthrough(),
    "context": multiquery_retriever | RunnableLambda(format_docs)
}
)
```

```python
parallel_chain.invoke("what is the medical condition mentioned here")
```

```python
chain= parallel_chain | prompt | chat_model2 | parser
```
